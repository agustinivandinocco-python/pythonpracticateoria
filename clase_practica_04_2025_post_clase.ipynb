### 9)

Supongamos que estamos evaluando otorgar prestamos con una tasa que varia en funcion del monto y del perfil de riesgo del beneficiario (bajo , medio y alto). Para montos mayores a 1.000.000 , la tasa sera del 5% sobre el capital. Para montos menores a 300.000 , la tasa sera del 12%. Para aquellos montos intermedios, la tasa estara en un 8%.

En cuanto al perfil, no se otorgara prestamos mayores a 1.000.000 para aquellos de riesgo alto. En caso de un perfil de riesgo bajo, se descontara un 1% a la tasa fijada sin importar el monto.

En caso de ser una solicitud aprobada, se guardara en un diccionario "aprobados" donde se usara el nombre como __key__, y el __monto + tasa__ como __value__.

en caso de no estar aprobada la solicitud, debera guardarse el nombre del solicitante en la lista "denegados"

Ademas se solicita:

- informar la cantidad de perfiles de riesgo alto (sin importar si el prestamo ha sido otorgado o no).
- sumar las cantidades de dinero otorgadas para perfiles de riesgo medio y bajo(no se debe sumar la tasa).

```python
solicitudes = {
    "Pepe": [200000 , "bajo"],
    "Juan": [3000000 , "medio"],
    "Victoria": [600000 , "medio"],
    "Agostina" : [1500000 , "alto"],
    "Julian" : [300000 , "alto"],
    "Francisco": [500000 , "alto"],
    "Pilar": [3000000 , "bajo"]
}

```

# Posible solucion ej 9

```python

solicitudes = {
    "Pepe": [200000 , "bajo"],
    "Juan": [3000000 , "medio"],
    "Victoria": [600000 , "medio"],
    "Agostina" : [1500000 , "alto"],
    "Julian" : [300000 , "alto"],
    "Francisco": [500000 , "alto"],
    "Pilar": [3000000 , "bajo"]
}


aprobados = {}
denegados = []
riesgo_alto = 0
suma_prestada = 0

for k , v in solicitudes.items():
    cliente = k
    monto = v[0]
    perfil_riesgo = v[1]
    BONIFICACION = 0.01

    #informo cantidad perfiles altos
    if perfil_riesgo == "alto":
        riesgo_alto+=1



    #denego alto + un millon
    if (monto >= 1000000) and (perfil_riesgo == "alto"):
        denegados.append(cliente)

    else:
        #aca ramifico en monto y evaluo riesgos
        if (monto >= 1000000):
            tasa = 0.05
            if perfil_riesgo =="bajo":
                tasa = tasa - BONIFICACION
            monto_final = monto * (1+tasa)


        elif monto >= 300000:
            tasa = 0.08
            if perfil_riesgo =="bajo":
                tasa = tasa - BONIFICACION
            monto_final = monto * (1+tasa)


        else:
            tasa = 0.12
            if perfil_riesgo =="bajo":
                tasa = tasa - BONIFICACION
            monto_final = monto * (1+tasa)

        aprobados[cliente] = round(monto_final,2)

        #sumo cantidad para medios y bajos
        if perfil_riesgo != "alto":
            suma_prestada += monto
```
